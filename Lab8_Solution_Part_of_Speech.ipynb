### Why this cell understates the tagger

The loop above calls `pos_tag([word])` once per word, so the tagger is handed a one-word
"sentence" every time. That throws away precisely the information a POS tagger exists to use.
NLTK's `pos_tag` is a perceptron tagger whose features include the previous two tags and the
surrounding words; given a single token it has none of them and falls back to what the word
looks like in isolation.

The effect is not subtle, and the next cell measures it on the lab's own sentence.

````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize

text = "I received my new phone broken from Amazon."
tokens = word_tokenize(text)

isolated = [pos_tag([word])[0][1] for word in tokens]   # as the cell above does it
in_context = [tag for _, tag in pos_tag(tokens)]        # the whole sentence at once

print(f"{'token':10} {'isolated':9} {'in context':10}")
for word, a, b in zip(tokens, isolated, in_context):
    flag = "  <-- differs" if a != b else ""
    print(f"{word:10} {a:9} {b:10}{flag}")

print()
print(f"{sum(a != b for a, b in zip(isolated, in_context))} of {len(tokens)} tags change")
````

Three of the nine tokens are tagged differently, and in every case the in-context tag is the
correct one:

| Token | Isolated | In context | What the isolated tag gets wrong |
|---|---|---|---|
| `received` | VBN | VBD | Past participle instead of past tense. Out of context `received` could be either; after `I` it can only be the simple past. |
| `broken` | NN | VBN | Tagged as a *noun*. Alone, `broken` is read as a thing; in `my new phone broken from Amazon` it is a participle describing the phone. |
| `Amazon` | NN | NNP | Common noun instead of proper noun. The capital letter alone is not enough, since the first word of any sentence is capitalised too — the tagger needs the position to know this one is a name. |

So the cell that is meant to introduce POS tagging demonstrates it with the context removed.
Calling `pos_tag(tokens)` on the whole list, as the lab does later in the adjective example,
is the correct usage and the one to copy.

This also explains why `broken` is the interesting word in the lab's chosen sentence. "I
received my new phone broken from Amazon" is a sentence where a human reader needs the word
order to understand who broke what, and the isolated tagger gets it wrong.

Two things in that output are worth stopping on.

`dog's` is split into `dog` and `'s`, and the `'s` is tagged POS, a possessive ending, rather
than being glued to the noun. spaCy tokenises on linguistic boundaries rather than whitespace,
which is why the token count is higher than the word count.

`jumped` gets VBD and `brown` gets JJ. The second one is a genuine decision rather than a
lookup: `brown` is also a common surname and a colour noun, and it is only the position between
a determiner-adjective sequence and a noun that makes it an adjective here. The comparison
cell further down shows NLTK getting exactly this word wrong.

In the English language, the same word can have different meanings, even within the same sentence.

For this reason, morphology is important. **spaCy** uses machine learning algorithms to best predict the use of a token in a sentence. Is *"I read books on NLP"* present or past tense?

````python
doc = nlp(u'I read books on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

````python
doc = nlp(u'I read a book on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

What tag will be assigned if you added **'always'**?

````python
doc = nlp(u'I always read a book on NLP.')
r = doc[2]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

### What the three `read` cells actually show

| Sentence | Tag for `read` | Reading |
|---|---|---|
| `I read books on NLP.` | VBP | present |
| `I read a book on NLP.` | VBD | past |
| `I always read a book on NLP.` | VBP | present |

The answer to the lab's question — is *"I read books on NLP"* present or past — is that spaCy
says present, and that it flips to past when `books` becomes `a book`.

It is worth being honest about how thin the evidence for that flip is. `read` is spelled the
same in both tenses, and both sentences are genuinely ambiguous in isolation: "I read a book on
NLP" is a perfectly good present-tense habitual statement. spaCy is not resolving the ambiguity
from meaning, it is going on a statistical cue — a bare plural object (`books`) patterns with
habitual present, a singular indefinite object (`a book`) patterns with narrative past in the
training data. Neither is a rule.

The third cell is the one that shows this most clearly. Adding `always` flips the tag back to
VBP, and `always` is a real cue: it forces a habitual reading, and habitual readings are
present tense. So the model does respond to a genuine signal when one is available — it just
also commits confidently when the only signal available is a weak one.

````python
# NLTK and spaCy on the identical sentence, tag by tag.
sentence = "The quick brown fox jumped over the lazy dog's back."

nltk_tags = pos_tag(word_tokenize(sentence))
spacy_doc = nlp(sentence)

print(f"{'token':10} {'nltk':6} {'spacy':6}")
nltk_lookup = {}
for word, tag in nltk_tags:
    nltk_lookup.setdefault(word, tag)

disagreements = 0
for token in spacy_doc:
    nltk_tag = nltk_lookup.get(token.text, "-")
    if nltk_tag not in ("-", token.tag_):
        disagreements += 1
        print(f"{token.text:10} {nltk_tag:6} {token.tag_:6}  <-- differs")

print()
print("tokens:", len(nltk_tags), "(nltk)", len(spacy_doc), "(spacy)")
print("disagreements on shared tokens:", disagreements)
````

One disagreement, and spaCy has it right: NLTK tags `brown` as NN, spaCy as JJ. In "The quick
brown fox", `brown` is an adjective modifying `fox`; NN would make it a noun, which would imply
a quite different sentence structure.

A single disagreement on a twelve-token sentence is not a basis for ranking the two taggers, and
it should not be read as one. What it does illustrate is where their error profiles differ:
NLTK's perceptron tagger works from the token sequence alone, while spaCy's tagger runs off
the same learned representation as its parser, so it tends to do better exactly where the
syntactic role of a word decides its category.

## Example: Find the adjective words in a sentance

````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize
````

````python
text =  "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."

print(text, end='\n\n')

for sent in sent_tokenize(text):
    tokens = word_tokenize(sent)
    part_of_speech = pos_tag(tokens)

    for i in part_of_speech:
        if 'JJ' in i:
            print(i)
    print('============================')
````

### The adjective finder misses adjectives

The filter in that cell is `if 'JJ' in i`, where `i` is a `(word, tag)` tuple. `in` on a tuple
tests whether an element *equals* `'JJ'`, so the condition is true only for the plain adjective
tag and never for `JJR` (comparative) or `JJS` (superlative). It reads like a substring test but
is not one.

That matters for this particular paragraph, as the next cell shows.

````python
from nltk.tokenize import sent_tokenize

# Same text as the cell above, condensed to the sentences that carry adjectives.
review = ("Mobile is very handy. Packed with good performances and nice camera. "
          "Battery is worst. I am hardly getting 5 hours screen on time.")

lab_version, fixed_version = [], []
for sentence in sent_tokenize(review):
    for pair in pos_tag(word_tokenize(sentence)):
        if 'JJ' in pair:                 # the lab's test: tuple membership
            lab_version.append(pair)
        if pair[1].startswith('JJ'):     # what was intended: any JJ* tag
            fixed_version.append(pair)

print("lab's filter  :", lab_version)
print("corrected     :", fixed_version)
print("missed        :", [p for p in fixed_version if p not in lab_version])
````

The lab's filter finds `handy`, `good` and `nice`, and misses `worst` (JJS).

The missed word is the one that matters most. This is a product review, the obvious reason to
extract adjectives from it is sentiment, and `worst` is the strongest sentiment-bearing word in
the paragraph — "Battery is worst" is the complaint. A filter that collects three positive
adjectives and silently drops the one strong negative does not just lose a word, it inverts the
summary of the review. `pair[1].startswith('JJ')` fixes it.

There is a second, quieter problem with `'JJ' in pair`: because it compares against both
elements of the tuple, a sentence containing the literal token `JJ` would match on the word
rather than the tag. That costs nothing on this text but it is the wrong test to copy.

### Task 1 — answers

`play` is token 12.

2. Fine-grained tag: **VB**
3. Meaning: **verb, base form**

VB rather than VBP is correct because `play` follows the infinitival `to` in "how to play it" —
an infinitive carries no tense, so the base-form tag is the right one. The full table from
part 4 is printed above.

Three rows of that table are worth picking out, because they show the coarse and fine-grained
tags disagreeing about what is interesting:

- `do` is AUX coarse but VBP fine. The fine-grained tagset has no auxiliary tag; it records
  tense and agreement and leaves the auxiliary/main-verb distinction to the coarse tag.
- `not` is PART coarse but RB fine — a particle in Universal Dependencies, an adverb in the
  Penn tagset. The same word, two defensible analyses from two different schemes.
- `to` is PART / TO, and `spacy.explain` gives `infinitival "to"`, correctly distinguishing it
  from the preposition `to`.

So the two tagsets are not a coarse and a fine view of one hierarchy. They disagree about
category in places, and `token.pos_` is not always recoverable from `token.tag_`.

ordinary English verb by usage. spaCy separates the two correctly, and note that it is not
simply going by the capital letter: the cue is the position after `Let us`, which can only be
followed by a verb.

`table` → `table` is older and purely syntactic. The same lowercase word, spelled identically,
is a noun after `on the` and a verb after `to`. Nothing about the word itself distinguishes
them; only the slot in the sentence does. "To table a motion" is also the less common sense by
a wide margin, so this is a case where the context has to overcome a strong prior toward the
furniture reading.

This is the practical reason POS tagging cannot be a dictionary lookup. A lookup table would
have to pick one answer per word and would be wrong on half of these four sentences.

### Task 3 — notes on the visualisation

The two clauses are parsed differently in a way the picture makes obvious.

`play` is VB, base form, because it sits under the infinitival `to` as the complement of
`like`. `hated` is VBD, past tense, and heads its own clause. So the same notional activity

`though` at the end is the most interesting token in the notebook, because it is tagged
differently here than in Task 1. Its dependency is `advmod` — an adverbial modifier of `hated`,

Both analyses are defensible for sentence-final `though`, and that is the point: it sits between
two categories, and spaCy resolved it one way in one sentence and the other way in another. The
dependency label stayed `advmod` in both. So the parse is the more stable description here, and
a tag that flips between two sentences with the same construction is a reminder that these

**What the tagging results show**

Context decides the tag, and the lab's examples are well chosen for showing it: `read` flips
VBP→VBD on nothing more than `books` becoming `a book`, and flips back when `always` is added.
The `always` case is the only one of the three where spaCy has a real linguistic cue to go on;
the other two are confident calls made on a weak statistical signal, which is worth remembering
